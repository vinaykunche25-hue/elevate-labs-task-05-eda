# Task 5 — Exploratory Data Analysis (EDA)

**Dataset:** synthetic customer-orders dataset created for practice; it is not real company data. The task brief permits any relevant dataset of your own choice.

Objective: extract insights using visual and statistical exploration with Pandas, Matplotlib and Seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Run from the notebooks/ folder
 df = pd.read_csv('../data/customer_orders_eda.csv')
df.head()
```

## 1. Inspect structure and statistics

```python
print('Shape:', df.shape)
df.info()
df.describe(include='all').T
```

## 2. Missing values and duplicates

```python
print('Missing values by column:\n', df.isna().sum())
print('Exact duplicate rows:', df.duplicated().sum())
```

## 3. Prepare data

Remove exact duplicate rows and fill missing Age and Satisfaction_Score with the median for this demonstration. For real data, investigate the cause of missingness before choosing a strategy.

```python
clean = df.drop_duplicates().copy()
for col in ['Age', 'Satisfaction_Score']:
    clean[col] = clean[col].fillna(clean[col].median())
print('Rows before:', len(df), '| after:', len(clean))
print('Missing values after handling:', int(clean.isna().sum().sum()))
```

## 4. Univariate analysis: annual spend

```python
clean['Annual_Spend'].describe()
```

```python
plt.figure(figsize=(8,4))
plt.hist(clean['Annual_Spend'], bins=30)
plt.xlabel('Annual spend'); plt.ylabel('Customers'); plt.title('Distribution of Annual Spend')
plt.show()
```

**Observation prompt:** Check whether the distribution is symmetric or right-skewed and whether any customers have unusually high spend.

## 5. Bivariate analysis

```python
clean.boxplot(column='Annual_Spend', by='Preferred_Category', rot=20, figsize=(8,4))
plt.title('Annual Spend by Preferred Category'); plt.suptitle('')
plt.show()
```

```python
clean.groupby('Region')['Annual_Spend'].agg(['sum','mean','count']).sort_values('sum', ascending=False)
```

```python
plt.figure(figsize=(8,4))
for label, group in clean.groupby('Loyalty_Status'):
    plt.scatter(group['Orders_Last_Year'], group['Annual_Spend'], label=label, alpha=.6)
plt.xlabel('Orders last year'); plt.ylabel('Annual spend'); plt.legend(); plt.show()
```

## 6. Multivariate analysis and correlation

```python
numeric_cols = ['Age','Orders_Last_Year','Annual_Spend','Returns','Satisfaction_Score']
corr = clean[numeric_cols].corr()
plt.figure(figsize=(8,5)); plt.imshow(corr.values, aspect='auto')
plt.xticks(range(len(corr.columns)), corr.columns, rotation=30, ha='right')
plt.yticks(range(len(corr.index)), corr.index)
for i in range(len(corr.index)):
    for j in range(len(corr.columns)):
        plt.text(j, i, f'{corr.iloc[i,j]:.2f}', ha='center', va='center')
plt.colorbar(); plt.title('Correlation Heatmap'); plt.show()
```

```python
from pandas.plotting import scatter_matrix
scatter_matrix(clean[numeric_cols], figsize=(10,8), diagonal='hist')
plt.show()
```

## 7. Summary of findings

- The data includes customer demographics, region, shopping channel, category, order frequency, annual spend, returns, loyalty status and satisfaction.
- Compare category and region summaries to identify segments with higher spend in this sample.
- Orders and spend can be compared using a scatterplot; correlation is association, not proof of causation.
- Missing values and duplicate rows were intentionally added to demonstrate data-quality checks.
